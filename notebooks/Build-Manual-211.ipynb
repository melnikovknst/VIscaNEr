# Build Manual 211

Runs the canonical joint YOLO once on every manually annotated image, aligns predictions to the reviewed target rectangles, creates paired bottle/label inputs, and packages a portable Kaggle dataset.

```python
from pathlib import Path
import json, subprocess, sys

start = Path.cwd().resolve()
PROJECT = next((p for p in (start, *start.parents) if (p / 'fusion_stage2').is_dir()), None)
if PROJECT is None:
    raise RuntimeError('Could not locate the VIscaNEr project root')

command = [sys.executable, '-u', '-m', 'fusion_stage2.build_manual_211_dataset']
subprocess.run(command, cwd=PROJECT, check=True)
```

```python
summary_path = PROJECT / 'datasets' / 'manual_211' / 'build_summary.json'
archive_path = PROJECT / 'datasets' / 'manual_211.zip'
summary = json.loads(summary_path.read_text(encoding='utf-8'))
print(json.dumps(summary, ensure_ascii=False, indent=2))
assert summary['complete'] and summary['rows'] == 211
assert archive_path.is_file()
print(f'READY | {archive_path} | {archive_path.stat().st_size / 2**20:.1f} MiB')
```
